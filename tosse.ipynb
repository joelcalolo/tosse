## 4. Construção do Modelo

Criando arquitetura CNN para classificação


````python
# Exemplo de construção do modelo
# Assumindo espectrogramas Mel de shape (128, 94) -> (128, 94, 1) para CNN

input_shape = (128, 94, 1)  # Ajuste conforme seus dados
num_classes = 3  # Normal, Bronquite, Pneumonia

classifier = CoughClassifier(
    input_shape=input_shape,
    num_classes=num_classes,
    model_type='cnn'  # ou 'mobilenet' para versão otimizada
)

model = classifier.get_model()
model.summary()
````
